*Time Based Analysis:*

1. Distribution of Duration (histogram)

```python
url = "https://raw.githubusercontent.com/ArchanaInsights/Datasets/main/marketing_campaign.csv"
df = pd.read_csv(url)

df['Date'] = pd.to_datetime(df['Date'], format='%d-%m-%Y')
df['Duration_Days'] = df['Duration'].str.extract('(\d+)').astype(int)
df['Month'] = df['Date'].dt.to_period('M')


print(df.columns.tolist())
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df['Duration_Days'], bins=10, kde=False, color='teal', edgecolor='black')
plt.title('Distribution of Campaign Duration')
plt.xlabel('Duration (days)')
plt.ylabel('Number of Campaigns')
plt.tight_layout()
plt.show()
```

2. Conversion_Rate over Date, by Company (line chart)

```python
conv_trend = df.groupby(['Date', 'Company'])['Conversion_Rate'].mean().reset_index()

plt.figure(figsize=(14,7))
sns.lineplot(data=conv_trend, x='Date', y='Conversion_Rate', hue='Company')
plt.title('Conversion Rate Trend Over Time by Company')
plt.xlabel('Date')
plt.ylabel('Average Conversion Rate')
plt.xticks(rotation=45)
plt.legend(title='Company', bbox_to_anchor=(1.02, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

3. Engagement_Score trend over Date (overall)



```python
engagement_trend = df.groupby('Date')['Engagement_Score'].mean().reset_index()

plt.figure(figsize=(14,6))
sns.lineplot(data=engagement_trend, x='Date', y='Engagement_Score', color='darkorange')
plt.title('Engagement Score Trend Over Time')
plt.xlabel('Date')
plt.ylabel('Average Engagement Score')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

**Geographic Analysis :**


1. Location with highest Acquisition_Cost (bar chart)

```python
location_cost = df.groupby('Location')['Acquisition_Cost'].mean().sort_values(ascending=False)
print(location_cost)

```

```python
plt.figure(figsize=(10,6))
sns.barplot(x=location_cost.index, y=location_cost.values, palette='mako')
plt.title('Average Acquisition Cost by Location')
plt.xlabel('Location')
plt.ylabel('Average Acquisition Cost')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

2. Conversion_Rate by Location, categorized by Target_Audience (grouped bar chart)

```python
loc_audience = df.groupby(['Location', 'Target_Audience'])['Conversion_Rate'].mean().reset_index()
print(loc_audience)
```

```python

plt.figure(figsize=(16,7))
sns.barplot(data=loc_audience, x='Location', y='Conversion_Rate', hue='Target_Audience')
plt.title('Conversion Rate by Location, Segmented by Target Audience')
plt.xlabel('Location')
plt.ylabel('Average Conversion Rate')
plt.xticks(rotation=45, ha='right')
plt.legend(title='Target Audience', bbox_to_anchor=(1.02, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

3. Proportion of ROI by Location (pie chart)

```python
location_roi = df.groupby('Location')['ROI'].sum().sort_values(ascending=False)

```

```python
plt.figure(figsize=(7,9))
plt.pie(location_roi, labels=location_roi.index, autopct='%1.1f%%', startangle=90,
        colors=sns.color_palette('Set3'))
plt.title('Proportion of Total ROI by Location')
plt.axis('equal')
plt.tight_layout()
plt.show()
```
